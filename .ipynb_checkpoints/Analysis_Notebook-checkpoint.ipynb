````python
### import libraries
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, median_absolute_error, r2_score, explained_variance_score, PredictionErrorDisplay
from sklearn.model_selection import KFold, cross_val_score

from nltk.sentiment import SentimentIntensityAnalyzer
````

````python
# read the dataset
df = pd.read_csv("data/test.csv")
df.shape
````

I have utilized some approaches to clean unrelated words within ```df['body']```, such as cleaning blank space, "\n" string,....

````python
df['body']
````

````python
## delete '\n' string inside the message
df['body'] = df['body'].str.replace('\n', ' ')
df['body'] = df['body'].str.replace(r'\s+', ' ', regex=True).str.strip()
df['body'] = df['body'].str.replace('=20', ' ', regex=False)
df['body'] = df['body'].str.replace('=01', '', regex=False)
````

````python
df['body']
````

### Task 1: Sentiment Labeling

````python
analyzer = SentimentIntensityAnalyzer()

sentiment_label = []
for sentence in df['body']:
    vs = analyzer.polarity_scores(sentence)
    score = vs['compound']
    if score >= 0.05:
        sentiment_label.append('Positive')
    elif score <= -0.05:
        sentiment_label.append('Negative')
    else:
        sentiment_label.append('Neutral')
````

````python
df['Sentiment'] = sentiment_label
df[['body','Sentiment']]
````

````python
print(f"The number of positive message(s): {(df['Sentiment'] == 'Positive').sum()}")
print(f"The number of neutral message(s): {(df['Sentiment'] == 'Neutral').sum()}")
print(f"The number of negative message(s): {(df['Sentiment'] == 'Negative').sum()}")
````

### Task 3: Employee Score Calculation

````python
"""
24 months 
10 employees
monthly score of 10 employees based on the message (df['body])

Approach: 
Column : Monthly Sentiment Score
Calculate the sentiment score based on df['Sentiment']
        if df['Sentiment'] == 'positive' -> df['Monthly Sentiment Score'] = 1
        else if df['Sentiment'] == 'negative --> df['Monthly Sentiment Score'] = -1
        else: df['Monthly Sentiment Score'] = 0 
Make a "MM-YYYY" column -> there would be 24 unique months
Calculate score of each employee each month = group by month -> then group by employee -> add df["Monthly Sentiment Score']
--> Ensure that the score resets at the beginning of each new month
"""
````

````python
# Set score for each sentiment
score = {'Positive': 1, 'Neutral': 0, 'Negative': -1}

# Assign a score to each message
df['Score'] = df['Sentiment'].map(score)
df.head(10)
````

````python
df['date'] = pd.to_datetime(df['date'])
df['Month'] = df['date'].dt.to_period('M')
df['Month']
````

````python
monthly_scores = df.groupby(['Month', 'from'])['Score'].sum()
monthly_scores = monthly_scores.reset_index()
monthly_scores = monthly_scores.rename(columns={'from': 'Employee', 'Score': 'Monthly Sentiment Score'})
monthly_scores.head(20)
````

````python
monthly_scores['Month'].unique()
# 2 years = 24 months (2010-2011)
````

````python
for employee in monthly_scores['Employee'].unique():
    print(f"{employee}")
    employee_name = monthly_scores[monthly_scores['Employee'] == employee]
    print(employee_name[['Month', 'Monthly Sentiment Score']].head(5))
````

### Task 4: Employee Ranking

````python
positive_scores = monthly_scores.sort_values(by=['Month', 'Monthly Sentiment Score', 'Employee'], ascending=[True, False, True])
top_three_positive = positive_scores.groupby('Month').head(3).reset_index(drop=True)
top_three_positive.set_index(['Month', 'Employee', 'Monthly Sentiment Score'])
````

````python
for month in top_three_positive['Month'].unique() :
    print(f'Month: {month}')
    top_positive = top_three_positive[top_three_positive['Month'] == month]
    print(top_positive[['Employee', 'Monthly Sentiment Score']].reset_index(drop=True))
````

````python
lowest_scores = monthly_scores.sort_values(by=['Month', 'Monthly Sentiment Score', 'Employee'], ascending=[True, True, True])
top_three_lowest = lowest_scores.groupby('Month').head(3).reset_index(drop=True)
top_three_lowest.set_index(['Month', 'Employee', 'Monthly Sentiment Score'])
````

````python
for month in top_three_lowest['Month'].unique() :
    print(f'Month: {month}')
    top_lowest = top_three_lowest[top_three_lowest['Month'] == month]
    print(top_lowest[['Employee', 'Monthly Sentiment Score']].reset_index(drop=True))
````

### Task 5: Flight Risk Identification

````python
"""
    The task requires us to find who has sent 4 or more negative mails ---> we filtered out the dataframe into only negative mails (df['Sentiment'] == 'Negative')
    We have to sort the dataframe by date and employee so it is easier for us to rolling 30-day period
"""

# Filter negative mails and sort the date
negative_df = df[df['Sentiment'] == 'Negative'].rename(columns={'from': 'Employee', 'date': 'Date'}).copy()
negative_df = negative_df.sort_values(['Employee', 'Date'])

# The 30-day period is rolling count of days, irrespective of months
rolling_30 = negative_df.set_index('Date').groupby('Employee')['Subject'].rolling('30D').count()
rolling_30 = rolling_30.reset_index(name='Neg Count')

# Extract a list of these employees 
flight_risk = rolling_30[rolling_30['Neg Count'] >= 4]['Employee'].unique()
for f in flight_risk :
    print(f"Flight Risk Employees: {f}")
````

### Task 6: Predictive Modeling

````python
df['Word Count'] = df['body'].str.split().str.len()

pred_df = df.groupby(['Month', 'from']).agg(Monthly_Score = ('Score', 'sum'),
                                            Message_Count = ('Score', 'size'),
                                            Monthly_Word_Count = ('Word Count', 'sum')).reset_index().rename(columns={'from': 'Employee'})
pred_df
````

````python
X = pred_df[['Message_Count', 'Monthly_Word_Count']]
y = pred_df['Monthly_Score']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=101)

print('Train:', X_train.shape, 'Val:', X_test.shape)
````

````python
model = LinearRegression()
kf_model = KFold(n_splits=5, shuffle=True, random_state=101)

neg_mse     = cross_val_score(model, X_train, y_train, cv=kf_model, scoring='neg_mean_squared_error')
rmse_scores = np.sqrt(-neg_mse)
r2_scores   = cross_val_score(model, X_train, y_train, cv=kf_model, scoring='r2')
mae_scores  = -cross_val_score(model, X_train, y_train, cv=kf_model, scoring='neg_mean_absolute_error')
````

````python
cv_table = pd.DataFrame({'Fold': range(1, 6),
                        'RMSE': rmse_scores,
                        'MAE' : mae_scores,
                        'R2'  : r2_scores}).set_index('Fold')
print(cv_table.round(4))
````

````python
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("R2  :", round(r2_score(y_test, y_pred),4))
print("RMSE:", round(np.sqrt(mean_squared_error(y_test, y_pred)),4))
print("MAE :", round(mean_absolute_error(y_test, y_pred),4))
````

````python
fig, ax = plt.subplots(figsize=(7, 5))
PredictionErrorDisplay.from_estimator(
    model, X_test, y_test, 
    kind="residual_vs_predicted",  # Specifies a Residual vs. Fitted plot
    ax=ax, 
    scatter_kwargs={"alpha": 0.6},
    line_kwargs={'color': 'red'}
)
ax.set_title("Residual vs. Fitted Plot")
plt.ylabel('Residuals')
plt.tight_layout()
plt.show()
````

````python
fig, ax = plt.subplots(figsize=(6, 6))
PredictionErrorDisplay.from_estimator(
    model, X_test, y_test,
    kind="actual_vs_predicted",   # <- the only change
    ax=ax,
    scatter_kwargs={"alpha": 0.6, "edgecolor": "k", "linewidth": 0.3},
    line_kwargs={'color': 'red', 'linestyle': "-"}
)
ax.set_title("Predicted vs Actual — Test set")
plt.tight_layout()
plt.show()
````
